```python
import pandas as pd
import pygris
```

```python
data = pd.read_csv("../../data/processed/CA_Fara_500_2019.csv")
los_angeles = data[data["CountyName"] == "Los Angeles"]
los_angeles_tracts = pygris.tracts(state="CA", county="Los Angeles", year=2019, cb=True)
```

```python
los_angeles_tracts["GEOID"] = los_angeles_tracts["GEOID"].astype(int)
```

```python
la = los_angeles_tracts.merge(los_angeles, how="inner", on="GEOID")
```
